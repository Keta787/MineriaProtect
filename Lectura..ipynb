```python
import pandas as pd

# 1. Cargar el archivo original sin modificarlo
df = pd.read_parquet('JobHop_v2_train.parquet')

print(f"Filas originales: {len(df)}")

# 2. Aplicar transformaciones en memoria
df['end_date'] = df['end_date'].fillna('Present')
df = df.dropna(subset=['start_date'])
df = df.drop_duplicates()
df = df.sort_values(by=['resume_id', 'start_date']).reset_index(drop=True)

# 3. Guardar el resultado en un archivo NUEVO (el original queda intacto)
df.to_parquet('JobHop_v2_train_limpio.parquet', index=False)

print(f"Filas después de limpiar: {len(df)}")
print("¡Archivo limpio guardado con éxito como 'JobHop_v2_train_limpio.parquet'!")
```

```python
# 1. Cargar el archivo limpio
df_limpio = pd.read_parquet('JobHop_v2_train_limpio.parquet')

# 2. Mostrar filas y columnas
print("--- DIMENSIONES DEL DATASET LIMPIO ---")
print(f"Filas: {df_limpio.shape[0]}")
print(f"Columnas: {df_limpio.shape[1]}")

# 3. Comprobar que no queden nulos críticos
print("\n--- VALORES NULOS RESTANTES ---")
print(df_limpio.isnull().sum())

# 4. Ver una muestra
display(df_limpio.head(5))
```
