# 4.3 · PCA

Solo variables numéricas estandarizadas. Binarias no se escalan.

```python
from pathlib import Path
import sys

def _carpeta_refactor() -> Path:
    for p in [Path.cwd(), *Path.cwd().parents]:
        if (p / 'lib' / 'config.py').exists(): return p
        if (p / 'notebooks' / 'refactor' / 'lib' / 'config.py').exists(): return p / 'notebooks' / 'refactor'
    raise RuntimeError('No encontrado')
RAIZ=_carpeta_refactor(); sys.path.insert(0,str(RAIZ))
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from lib import carga, categorias, periodo
%matplotlib inline
print('ok')
```

```python
df = carga.leer(config.CON_OUTLIERS)
df['duracion']=pd.to_numeric(periodo.duracion_observada(df), errors='coerce')
df['nivel_ordinal']=pd.to_numeric(categorias.nivel_ordinal(df), errors='coerce')
Xn=df[['nivel_ordinal','n_skills_essential','veces_ese_oficio','duracion']].copy()
Xn['nivel_ordinal']=Xn['nivel_ordinal'].fillna(Xn['nivel_ordinal'].median())
for c in Xn.columns: Xn[c]=pd.to_numeric(Xn[c], errors='coerce').fillna(Xn[c].median())
scaler=StandardScaler(); Xn_s=scaler.fit_transform(Xn)
pca=PCA(); Xp=pca.fit_transform(Xn_s)
var=pca.explained_variance_ratio_
plt.bar(range(1,len(var)+1),var); plt.plot(range(1,len(var)+1),np.cumsum(var),marker='o'); plt.title('Varianza explicada'); plt.show()
print(np.cumsum(var).round(4))
```
